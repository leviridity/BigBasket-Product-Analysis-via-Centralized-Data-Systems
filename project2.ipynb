```python
# Convert categorical columns to object
bb['product'] = bb['product'].fillna('Unknown')
bb['brand'] = bb['brand'].fillna('Unknown')
bb['description'] = bb['description'].fillna('No description available')
bb.head()
```

```python
# Check for columns with missing values
missing_columns = bb.columns[bb.isnull().any()]
print("Columns with missing values:", missing_columns)
```

```python
# Standardize case
categorical_columns = bb.select_dtypes(include=['object']).columns
bb[categorical_columns] = bb[categorical_columns].apply(lambda x: x.str.lower())
bb.head()
```

```python
bb.tail()
```

```python
# Strip whitespace - Remove leading and trailing whitespaces in text fields.
bb[categorical_columns] = bb[categorical_columns].apply(lambda x: x.str.strip())
bb.head()
```

```python
bb.tail()
```

```python
# Data Type Conversion
bb.dtypes
```

```python
bb['product'] = bb['product'].astype('object')
bb['category'] = bb['category'].astype('category')
bb['sub_category'] = bb['sub_category'].astype('category')
bb['brand'] = bb['brand'].astype('category')
bb['type'] = bb['type'].astype('category')
bb['sale_price'] = bb['sale_price'].astype(float)
bb['market_price'] = bb['market_price'].astype(float)
bb['rating'] = bb['rating'].astype(float)
bb['description'] = bb['description'].astype(str)
```

```python
bb.dtypes
```

```python
# Remove duplicates
bb.drop_duplicates()
bb.head()
```

```python
bb.tail()
```

```python
# Validation: Flag rows where sale_price >= market_price
bb['invalid_price'] = bb['sale_price'] >= bb['market_price']
invalid_rows = bb[bb['invalid_price'] == True]
```

```python
invalid_rows.head()
```

```python
invalid_rows.tail()
```

```python
bb
```

```python
# Discount/Price Difference
bb['market_price'] = pd.to_numeric(bb['market_price'], errors='coerce')
bb['sale_price'] = pd.to_numeric(bb['sale_price'], errors='coerce')

bb['price_difference'] = bb['market_price'] - bb['sale_price']
bb['price_difference'] = bb['price_difference'].apply(lambda x: f"{x:.2f}" if pd.notnull(x) else "")
```

```python
bb.head()
```

```python
bb.tail()
```

```python
# Price Range Category
def price_category(price):
    if price < 100:
        return 'Low'
    elif 100 <= price < 500:
        return 'Medium'
    else:
        return 'High'

bb['price_category'] = bb['sale_price'].apply(price_category)
bb.head()
```

```python
bb.tail()
```

```python
# Demand Category based on Rating
def demand_category(rating):
    if rating >= 4.5:
        return 'High Demand'
    elif rating >= 3.0:
        return 'Medium Demand'
    else:
        return 'Low Demand'

bb['demand_tier'] = bb['rating'].apply(demand_category)
bb.head()
```

```python
bb.tail()
```

```python
bb.to_csv('bigbasket_transformed.csv', index=False)
```

```python
# Insights and Findings
# Product Category Distribution
import matplotlib.pyplot as plt
category_counts = bb['category'].value_counts()
category_counts.plot(kind='bar', figsize=(10, 6), color='pink')
plt.title('Product Category Distribution')
plt.xlabel('Categories')
plt.ylabel('Number of Products')
plt.show()
```

```python
# Top 10 Bestselling Products by Sale Price
top_bestselling = bb.sort_values(by='sale_price', ascending=False).head(10)
top_bestselling.plot(kind='barh', x='product', y='sale_price', figsize=(10, 6), color='pink')
plt.title('Top 10 Bestselling Products by Sale Price')
plt.xlabel('Sale Price')
plt.ylabel('Product Name')
plt.show()
```
